# Graph

1. Biểu diễn đồ thị bằng ma trận kề.    

Đồ thị có thể được biểu diễn dưới dạng ma trận kề 
Với 1 ma trận có n đỉnh thì mà trận sẽ có dạng n*n

```python
n=int(input("Nhập số đỉnh của ma trận"))
mt=[]
print("Nhập ma trận kề của các đỉnh")
#Nhap ma tran
for x in range(n):
    for y in range(n):
        cot=[int(i) for i in input().split()]
        mt.append(cot)
#in ma tran ke
    for row in mt:
        for col in row:
            print(col,end=" ")
    print()
```
